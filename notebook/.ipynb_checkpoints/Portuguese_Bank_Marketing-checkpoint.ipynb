# 7. OUTLIER DETECTION

Outlier detection is performed to identify unusually
high or low numerical observations.

The Interquartile Range (IQR) method is used to
identify potential outliers in numerical features.


```python

# IQR Based Outlier Detection

outlier_columns = [
    'age',
    'campaign',
    'previous',
    'pdays',
    'duration'
]

outlier_results = []

for col in outlier_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ]

    outlier_results.append({
        'Feature': col,
        'Lower Limit': lower_limit,
        'Upper Limit': upper_limit,
        'Outlier Count': len(outliers),
        'Outlier Percentage': len(outliers) / len(df) * 100
    })

outlier_df = pd.DataFrame(outlier_results)

display(outlier_df.round(2))
```

### Prepare Data for Machine Learning


# 9. DATA PREPROCESSING

Data preprocessing prepares the dataset for machine learning.

The target variable is separated from the independent variables.

The duration feature is excluded because it is only known
after the marketing call has been completed.

This prevents target leakage and supports realistic
customer subscription prediction.


### Separate X and y

```python

# Separate Independent and Dependent Variables

# Target Variable
y = df_model['y']

# Drop target and leakage columns
X = df_model.drop(
    columns=['y', 'y_encoded', 'duration'],
    errors='ignore'
)

print("Features Shape:", X.shape)
print("Target Shape:", y.shape)

print("\nTarget Distribution:")
display(y.value_counts())

print("\nFeatures:")
print(X.columns.tolist())
```

### Identify Numerical and Categorical Columns

```python

# Identify Numerical and Categorical Features

numerical_features = X.select_dtypes(
    include=np.number
).columns.tolist()

categorical_features = X.select_dtypes(
    include='object'
).columns.tolist()

print("Numerical Features:")
print(numerical_features)

print("\nCategorical Features:")
print(categorical_features)

print("\nTotal Numerical Features:",
      len(numerical_features))

print("Total Categorical Features:",
      len(categorical_features))
```

### Train-Test Split

```python

# Train Test Split

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Data Shape:", X_train.shape)
print("Testing Data Shape:", X_test.shape)

print("\nTraining Target Distribution:")
display(y_train.value_counts(normalize=True).round(3))

print("\nTesting Target Distribution:")
display(y_test.value_counts(normalize=True).round(3))
```


### Data Preprocessing Summary

1. Independent and dependent variables were separated.
2. Duration was excluded to prevent target leakage.
3. Numerical and categorical features were identified.
4. The dataset was divided into training and testing sets.
5. Stratified sampling was used to preserve target class proportions.

The prepared data will be used to build and evaluate
multiple classification models.



# 10. MACHINE LEARNING MODEL BUILDING

Multiple machine learning classification algorithms
are implemented to predict whether a customer will
subscribe to a bank term deposit.

The models are evaluated using multiple classification
metrics to identify the most suitable model for production.


```python

# Import Machine Learning Libraries

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    classification_report,
    confusion_matrix
)

from xgboost import XGBClassifier

print("Machine Learning Libraries Imported Successfully!")
```

### Data Preprocessing Pipeline

```python

# Numerical Preprocessing

numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])

# Categorical Preprocessing

categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore'))
])

# Combine Both Transformers

preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numerical_features),
        ('cat', categorical_transformer, categorical_features)
    ]
)

print("Preprocessing Pipeline Created Successfully!")
```

### Define Multiple Machine Learning Models

```python

# Define Machine Learning Models

class_ratio = (
    (y_train == 'no').sum() /
    (y_train == 'yes').sum()
)

models = {
    'Logistic Regression': LogisticRegression(
        max_iter=2000,
        class_weight='balanced',
        random_state=42
    ),

    'Decision Tree': DecisionTreeClassifier(
        class_weight='balanced',
        random_state=42
    ),

    'Random Forest': RandomForestClassifier(
        n_estimators=200,
        class_weight='balanced_subsample',
        random_state=42,
        n_jobs=-1
    ),

    'XGBoost': XGBClassifier(
        n_estimators=200,
        max_depth=4,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        scale_pos_weight=class_ratio,
        eval_metric='logloss',
        random_state=42,
        n_jobs=-1
    )
}

print("Total Models:", len(models))

for name in models:
    print(name)
```

### Train All Models

## Target Variable Change

```python

# Convert Target Variable into Numerical Format

y = df_model['y'].map({
    'no': 0,
    'yes': 1
})

print("Target Variable Converted Successfully!")

print(y.value_counts())
```

### Train Test Split Again

```python

# Train Test Split

from sklearn.model_selection import train_test_split

X = df_model.drop(
    columns=['y', 'y_encoded', 'duration'],
    errors='ignore'
)

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training Shape:", X_train.shape)
print("Testing Shape:", X_test.shape)

print("\nTraining Target:")
print(y_train.value_counts())
```

### Train All Models Again

```python

# Train All Machine Learning Models

trained_models = {}

for name, model in models.items():

    print("\nTraining:", name)

    pipeline = Pipeline(steps=[
        ('preprocessor', preprocessor),
        ('classifier', model)
    ])

    pipeline.fit(X_train, y_train)

    trained_models[name] = pipeline

    print(name, "Training Completed!")

print("\nAll Models Trained Successfully!")
```

### Update Evaluation Code

```python

# Model Evaluation

results = []

for name, model in trained_models.items():

    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1]

    results.append({
        'Model': name,
        'Accuracy': accuracy_score(y_test, y_pred),
        'Precision': precision_score(
            y_test, y_pred, pos_label=1
        ),
        'Recall': recall_score(
            y_test, y_pred, pos_label=1
        ),
        'F1 Score': f1_score(
            y_test, y_pred, pos_label=1
        ),
        'ROC AUC': roc_auc_score(y_test, y_prob),
        'PR AUC': average_precision_score(y_test, y_prob)
    })

model_comparison = pd.DataFrame(results)

model_comparison = model_comparison.sort_values(
    by='PR AUC',
    ascending=False
).reset_index(drop=True)

display(model_comparison.round(4))
```

### Compare All Models

```python

# Model Performance Comparison

display(
    model_comparison.style
    .format({
        'Accuracy': '{:.4f}',
        'Precision': '{:.4f}',
        'Recall': '{:.4f}',
        'F1 Score': '{:.4f}',
        'ROC AUC': '{:.4f}',
        'PR AUC': '{:.4f}'
    })
    .highlight_max(
        subset=[
            'Accuracy', 'Precision', 'Recall',
            'F1 Score', 'ROC AUC', 'PR AUC'
        ],
        color='lightgreen'
    )
)
```

### Identify Best Model

```python

# Select Best Model Based on PR-AUC

best_model_name = model_comparison.iloc[0]['Model']

best_model = trained_models[best_model_name]

print("BEST MODEL:", best_model_name)

print("\nBest Model Performance:")

display(
    model_comparison[
        model_comparison['Model'] == best_model_name
    ]
)
```

### ROC Curve Comparison

```python

# ROC Curve Comparison

from sklearn.metrics import roc_curve, auc

plt.figure(figsize=(10, 7))

for name, model in trained_models.items():

    y_probability = model.predict_proba(X_test)[:, 1]

    fpr, tpr, thresholds = roc_curve(
        y_test,
        y_probability
    )

    roc_auc = auc(fpr, tpr)

    plt.plot(
        fpr,
        tpr,
        label=f'{name} (AUC = {roc_auc:.3f})'
    )

plt.plot([0, 1], [0, 1], 'k--')

plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')

plt.title('ROC Curve Comparison')

plt.legend()
plt.grid()
plt.show()
```

### Precision Recall Curve

```python

# Precision Recall Curve

from sklearn.metrics import precision_recall_curve

plt.figure(figsize=(10, 7))

for name, model in trained_models.items():

    y_probability = model.predict_proba(X_test)[:, 1]

    precision, recall, thresholds = precision_recall_curve(
        y_test,
        y_probability
    )

    plt.plot(
        recall,
        precision,
        label=name
    )

plt.xlabel('Recall')
plt.ylabel('Precision')

plt.title('Precision Recall Curve Comparison')

plt.legend()
plt.grid()
plt.show()
```

# 12. HYPERPARAMETER AND THRESHOLD OPTIMIZATION

Threshold optimization is performed to improve
the identification of potential term deposit subscribers.

The validation dataset is used to select a suitable
classification threshold without tuning directly
on the test dataset.


```python

from sklearn.base import clone

# Split Training Data into Training and Validation

X_fit, X_val, y_fit, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.20,
    random_state=42,
    stratify=y_train
)

# Create XGBoost Pipeline

xgb_pipeline = Pipeline(steps=[
    ('preprocessor', clone(preprocessor)),
    ('classifier', clone(models['XGBoost']))
])

# Train on training subset
xgb_pipeline.fit(X_fit, y_fit)

# Validation probabilities
val_prob = xgb_pipeline.predict_proba(X_val)[:, 1]

print("Validation Training Completed!")
print("Validation Records:", len(X_val))
```

### Find Best Threshold

```python

# Threshold Optimization

threshold_results = []

for threshold in np.arange(0.20, 0.81, 0.05):

    val_pred = (val_prob >= threshold).astype(int)

    threshold_results.append({
        'Threshold': threshold,
        'Precision': precision_score(
            y_val, val_pred, zero_division=0
        ),
        'Recall': recall_score(
            y_val, val_pred, zero_division=0
        ),
        'F1 Score': f1_score(
            y_val, val_pred, zero_division=0
        )
    })

threshold_df = pd.DataFrame(threshold_results)

display(threshold_df.round(4))

best_threshold = threshold_df.loc[
    threshold_df['F1 Score'].idxmax(),
    'Threshold'
]

print("Best Threshold:", best_threshold)
```

### Threshold Visualization

```python

plt.figure(figsize=(10, 5))

plt.plot(
    threshold_df['Threshold'],
    threshold_df['Precision'],
    marker='o',
    label='Precision'
)

plt.plot(
    threshold_df['Threshold'],
    threshold_df['Recall'],
    marker='o',
    label='Recall'
)

plt.plot(
    threshold_df['Threshold'],
    threshold_df['F1 Score'],
    marker='o',
    label='F1 Score'
)

plt.title('Threshold Optimization - XGBoost')
plt.xlabel('Classification Threshold')
plt.ylabel('Score')

plt.legend()
plt.grid()
plt.show()
```

### Final XGBoost Model

```python

# Train Final XGBoost Model

best_model = Pipeline(steps=[
    ('preprocessor', clone(preprocessor)),
    ('classifier', clone(models['XGBoost']))
])

best_model.fit(X_train, y_train)

# Test Prediction
test_prob = best_model.predict_proba(X_test)[:, 1]

test_pred = (test_prob >= best_threshold).astype(int)

print("Final XGBoost Model Trained Successfully!")
print("Selected Threshold:", best_threshold)

print("\nFinal Test Performance:")

print(classification_report(
    y_test,
    test_pred,
    target_names=['No Subscription', 'Subscribed']
))
```

### Final Confusion Matrix

```python

cm = confusion_matrix(y_test, test_pred)

plt.figure(figsize=(7, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=['No', 'Yes'],
    yticklabels=['No', 'Yes']
)

plt.title('Final XGBoost Confusion Matrix')
plt.xlabel('Predicted')
plt.ylabel('Actual')

plt.show()
```

### Feature Importance Analysis

```python

# FEATURE IMPORTANCE ANALYSIS

# Get trained XGBoost classifier
xgb_model = best_model.named_steps['classifier']

# Get feature names after preprocessing
feature_names = best_model.named_steps[
    'preprocessor'
].get_feature_names_out()

# Get feature importance values
importance_values = xgb_model.feature_importances_

# Create DataFrame
feature_importance_df = pd.DataFrame({
    'Feature': feature_names,
    'Importance': importance_values
})

# Sort features
feature_importance_df = feature_importance_df.sort_values(
    by='Importance',
    ascending=False
).reset_index(drop=True)

print("Top 20 Important Features")

display(feature_importance_df.head(20).round(4))
```

### Challenges Faced Report


# 15. CHALLENGES FACED AND SOLUTIONS

## 1. Dataset Imbalance

Challenge:
The target variable contains more customers who did not
subscribe compared to customers who subscribed.

Solution:
Used stratified train-test splitting and evaluated
Precision, Recall, F1-score and PR-AUC.

## 2. Missing and Unknown Values

Challenge:
Several categorical features contain unknown values.

Solution:
Handled unknown categories through preprocessing
and used suitable encoding techniques.

## 3. Categorical Variables

Challenge:
Machine learning algorithms require numerical
representations of categorical variables.

Solution:
Applied One-Hot Encoding to categorical features.

## 4. Outliers

Challenge:
Numerical features contained extreme values.

Solution:
Used the IQR method to identify outliers and
analysed their distribution.

## 5. Data Leakage

Challenge:
The duration variable is only known after a call.
Using it before a call would create unrealistic predictions.

Solution:
Excluded duration from the predictive model.

## 6. Model Performance

Challenge:
Different algorithms produced different performance
results, especially in identifying subscribers.

Solution:
Compared Logistic Regression, Decision Tree,
Random Forest and XGBoost using multiple metrics.

## 7. Threshold Selection

Challenge:
The default classification threshold may not
provide the best balance between precision and recall.

Solution:
Performed threshold optimization using validation data.

## Conclusion

The challenges were addressed using appropriate
data preprocessing, exploratory analysis,
model evaluation and optimization techniques.
